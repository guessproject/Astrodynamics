# Low Thrust Transfer Orbits

## The Lagrange planetary equations

For continuous, low thrust maneuvers the most useful form is the **Gauss form of the Lagrange planetary equations**; here thrust is introduced directly as an acceleration in the local orbital frame. Low-Thrust Maneuvers for the Efficient Correction of Orbital Elements states this explicitly and defines the thrust using an **in-plane angle** $\alpha$ and an **out-of-plane angle** $\beta$.

## Thrust-angle convention

, where:

* $\hat{r}$: radial, outward from the central body
* $\hat{c}$: circumferential, in the orbital plane and orthogonal to $\hat{r}$
* $\hat{n}$: orbit normal, along angular momentum

The thrust steering angles are defined as:

* $\alpha$ **in-plane thrust angle**, positive from the circumferential direction toward radial-outward
* $\beta$: **out-of-plane thrust angle**, positive from the orbital plane toward the angular-momentum direction

Defining $f$ as the thrust acceleration magnitude, the components are

$$
f_R = f\cos\beta \sin\alpha,\qquad
f_T = f\cos\beta \cos\alpha,\qquad
f_N = f\sin\beta
$$


## Gauss/Lagrange planetary equations as functions of $\alpha,\beta$

Let

* $a$: semimajor axis
* $e$: eccentricity
* $i$: inclination
* $\Omega$: RAAN
* $\omega$: argument of perigee
* $\theta$: true anomaly
* $u = \omega + \theta$: argument of latitude
* $p=a(1-e^2)$: semilatus rectum
* $r=\dfrac{p}{1+e\cos\theta}$: orbital radius
* $h=\sqrt{\mu p}$: specific angular momentum

Then the element rates become

$$\dot a=\frac{2a^2}{h}\,f\cos\beta\left(e\sin\theta \sin\alpha + \frac{p}{r}\cos\alpha\right)$$

$$\dot e=\frac{1}{h}\,f\cos\beta\left[p\sin\theta \sin\alpha +\big((p+r)\cos\theta + re\big)\cos\alpha\right]$$

$$\dot i=\frac{r}{h}\,f\sin\beta \cos u$$

$$\dot\Omega=\frac{r}{h\sin i}\,f\sin\beta \sin u$$

$$\dot\omega=\frac{1}{he}\,f\cos\beta\left[-p\cos\theta \sin\alpha + (p+r)\sin\theta \cos\alpha\right]-\frac{r\cos i}{h\sin i}\,f\sin\beta \sin u$$

These are the thrust-angle form of the Gauss/Lagrange planetary equations given by substituting the $(\alpha,\beta)$ thrust decomposition into the Gauss equations. These equations give, for each orbital element, **the analytical relation between thrust angles $(\alpha,\beta)$ and the instantaneous orbital-element variation**.

From these equations:

* **In-plane thrust** $(\beta=0)$ drives mainly $a,e,\omega$. 
* **Out-of-plane thrust** $(\alpha$ irrelevant, $\beta=\pm \pi/2)$ drives mainly $i,\Omega$.
* The maximum $\dot a$$ and $\dot e$ occur for **planar thrusting** $(\beta=0)$, whereas $i$ and $\Omega$ depend only on the out-of-plane thrust term.
